Eine **Klasse** ist ein *Bauplan* – sie beschreibt, welche Eigenschaften (Attribute) und Fähigkeiten (Methoden) etwas hat. Ein **Objekt** (auch "Instanz" genannt) ist ein konkretes Ding, das nach diesem Bauplan gebaut wurde.

#### Klasse "Hund"

Bauplan: hat einen Namen, eine Rasse, kann bellen()

#### Objekt 1

Name: "Bello"\
Rasse: "Labrador"

#### Objekt 2

Name: "Rex"\
Rasse: "Dackel"

Vergleich zu dem, was ihr schon kennt: Ein `dict` könnte Daten wie `{"name": "Bello"}` speichern – aber eine Klasse kann zusätzlich **Funktionen (Methoden)** direkt mitliefern, die zu diesen Daten gehören.

Grundgerüst einer Klasse:

```
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        print(f"{self.name} sagt: Wuff!")
```

- `class Hund:` – definiert den Bauplan
- `__init__` – wird automatisch beim Erstellen eines Objekts aufgerufen (Konstruktor)
- `self` – verweist auf das jeweilige Objekt selbst
- `self.name = name` – speichert ein Attribut am Objekt
- `def bellen(self):` – eine Methode, also eine Funktion, die zur Klasse gehört

So wird ein Objekt erzeugt und benutzt:

```
bello = Hund("Bello", "Labrador")
bello.bellen()
print(bello.name)

# Ausgabe:
# Bello sagt: Wuff!
# Bello
```

Man kann beliebig viele Objekte aus derselben Klasse erzeugen – jedes hat seine eigenen Attribute:

```
rex = Hund("Rex", "Dackel")
rex.bellen()
# Rex sagt: Wuff!
```

Zweites Beispiel: eine Klasse, deren Methoden den **Zustand des Objekts verändern** – nicht nur ausgeben.

```
class Konto:
    def __init__(self, inhaber, kontostand=0):
        self.inhaber = inhaber
        self.kontostand = kontostand

    def einzahlen(self, betrag):
        self.kontostand += betrag
        print(f"{betrag}€ eingezahlt. Neuer Stand: {self.kontostand}€")

    def abheben(self, betrag):
        if betrag > self.kontostand:
            print("Nicht genug Guthaben!")
        else:
            self.kontostand -= betrag
            print(f"{betrag}€ abgehoben. Neuer Stand: {self.kontostand}€")
```

Verwendung:

```
konto = Konto("Anna", 100)
konto.einzahlen(50)
konto.abheben(30)

# Ausgabe:
# 50€ eingezahlt. Neuer Stand: 150€
# 30€ abgehoben. Neuer Stand: 120€
```

Wichtiger Unterschied zum Hund-Beispiel: Hier ändert eine Methode (`self.kontostand += betrag`) ein Attribut dauerhaft – das Objekt "merkt sich" seinen neuen Zustand.

### Warum nicht einfach dict + Funktionen?

Theoretisch könnte man Daten in einem `dict` speichern und separate Funktionen schreiben, die dieses dict als Parameter bekommen. Das funktioniert bei kleinen Programmen, wird aber schnell unübersichtlich: Daten und die Funktionen, die zu ihnen gehören, liegen getrennt voneinander im Code, und jede Funktion muss das dict als Parameter durchreichen.

Eine Klasse bündelt **Daten (Attribute) und Verhalten (Methoden)** an einem Ort. Das Objekt trägt sein Verhalten quasi "in sich" – `hund.bellen()` statt `bellen(hund_dict)`.

**Was daran wichtig ist:**

- **Struktur:** Größere Programme werden in sinnvolle, in sich geschlossene Bausteine aufgeteilt.
- **Wiederverwendbarkeit:** Eine Klasse einmal schreiben, beliebig viele Objekte davon erzeugen.
- **Wartbarkeit:** Ändert sich, wie ein Hund bellt, muss das nur an einer Stelle (in der Methode) angepasst werden.
- **Kapselung:** Ein Objekt kann seine eigenen Daten konsistent halten, statt dass jede Funktion im Programm das dict beliebig verändert.

### Aufgabe: Klasse Bankkonto

Schreibt eine Klasse `Bankkonto` mit folgenden Anforderungen:

**Attribute:**
- `inhaber` – der Name des Kontoinhabers
- `kontostand` – der aktuelle Kontostand (wird beim Erstellen übergeben)

**Methoden:**
1. `__init__(self, inhaber, kontostand)` – initialisiert die beiden Attribute mit den übergebenen Werten.
2. `zeige_kontostand(self)` – gibt den Kontostand im Format aus:
   `"Kontostand von {inhaber}: {kontostand} €"`
3. `einzahlen(self, betrag)` – erhöht `kontostand` um `betrag`.
4. `abheben(self, betrag)`:
   - Ist `betrag` größer als der aktuelle `kontostand`, soll `"Nicht genügend Guthaben vorhanden!"` ausgegeben werden.
   - Andernfalls wird `kontostand` um `betrag` verringert.

**Testet eure Klasse:**
- Erstellt ein Bankkonto-Objekt mit einem Inhaber und einem Startguthaben.
- Zahlt einen Betrag ein und zeigt den Kontostand an.
- Hebt einen Betrag ab, der größer ist als das Guthaben, und prüft, ob die Fehlermeldung erscheint.
- Hebt anschließend einen gültigen Betrag ab und zeigt den Kontostand erneut an.

````python
ergebnis = []
for x in range(20):
    if x % 2 == 0:
        ergebnis.append(x ** 2)

ergebnis = [x ** 2 for x in range(20) if x % 2 == 0]
````
