# Kapitel 4: Funktionen

Wiederverwendbare Codeblöcke mit Parametern und Rückgabewerten.

## 1. Grundgerüst

```python
def begruessung(name):
    return f"Hallo {name}!"

print(begruessung("Anna"))
```

## 2. Mehrere Parameter & Default-Werte

```python
def rabatt(preis, prozent=10):
    return preis * (1 - prozent / 100)

print(rabatt(100))       # nutzt den Default-Wert 10
print(rabatt(100, 25))   # überschreibt den Default-Wert
```

## 3. Mehrere Rückgabewerte

```python
def min_max(zahlen):
    return min(zahlen), max(zahlen)

kleinstes, groesstes = min_max([4, 8, 15, 16, 23, 42])
print(kleinstes, groesstes)
```

## 4. `*args` und `**kwargs`

Für eine variable Anzahl an Argumenten.

```python
def summe(*zahlen):
    return sum(zahlen)

print(summe(1, 2, 3))
print(summe(1, 2, 3, 4, 5))

def profil(**infos):
    for key, wert in infos.items():
        print(key, "->", wert)

profil(name="Anna", alter=23)
```

## 5. Funktionen als Parameter

```python
def wende_an(funktion, wert):
    return funktion(wert)

def verdoppeln(x):
    return x * 2

print(wende_an(verdoppeln, 5))
print(wende_an(lambda x: x ** 2, 5))   # anonyme Funktion
```
