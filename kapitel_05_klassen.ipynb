# Kapitel 5: Klassen & OOP

Eine Klasse ist ein Bauplan – sie beschreibt Attribute (Daten) und Methoden
(Verhalten). Ein Objekt ist eine konkrete Instanz dieses Bauplans.

## 1. Grundgerüst

```python
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        print(f"{self.name} sagt: Wuff!")

bello = Hund("Bello", "Labrador")
bello.bellen()
print(bello.name)
```

## 2. Mehrere Objekte, gleicher Bauplan

```python
rex = Hund("Rex", "Dackel")
rex.bellen()
```

## 3. Methoden, die den Zustand ändern

```python
class Konto:
    def __init__(self, inhaber, kontostand=0):
        self.inhaber = inhaber
        self.kontostand = kontostand

    def einzahlen(self, betrag):
        self.kontostand += betrag
        print(f"{betrag}€ eingezahlt. Neuer Stand: {self.kontostand}€")

    def abheben(self, betrag):
        if betrag > self.kontostand:
            print("Nicht genug Guthaben!")
        else:
            self.kontostand -= betrag
            print(f"{betrag}€ abgehoben. Neuer Stand: {self.kontostand}€")

konto = Konto("Anna", 100)
konto.einzahlen(50)
konto.abheben(30)
```

## 4. Objekte in Listen

Python kennt keine "Arrays" im klassischen Sinn – dafür verwendet man `list`,
und Listen können genauso gut Objekte enthalten.

```python
hunde = [Hund("Bello", "Labrador"), Hund("Rex", "Dackel"), Hund("Luna", "Pudel")]

for hund in hunde:
    hund.bellen()
```

## 5. `__del__` – der Destruktor

Das Gegenstück zu `__init__`: wird automatisch aufgerufen, wenn ein Objekt
gelöscht wird.

```python
class Auto:
    def __init__(self, marke, tankfuellung=40):
        self.marke = marke
        self.tankfuellung = tankfuellung

    def fahren(self, liter):
        self.tankfuellung -= liter
        print(f"{self.marke}: noch {self.tankfuellung}L im Tank")

    def __del__(self):
        print(f"{self.marke} wurde verschrottet.")

mein_auto = Auto("VW")
mein_auto.fahren(10)
del mein_auto
```

## Warum Klassen statt dict + Funktionen?

Ein `dict` speichert nur Daten. Eine Klasse bündelt Daten **und** die Funktionen,
die zu diesen Daten gehören, an einem Ort: `hund.bellen()` statt
`bellen(hund_dict)`. Das macht größere Programme strukturierter, leichter
wiederverwendbar und wartbarer.
