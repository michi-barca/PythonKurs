# Kapitel 3: Datenstrukturen

Listen, Tupel, Dictionaries, Sets – und List Comprehensions.

## 1. Listen

```python
zahlen = [4, 8, 15, 16, 23, 42]
print(zahlen[0], zahlen[-1])
print(zahlen[1:3])

zahlen.append(100)
zahlen.remove(8)
print(zahlen)
print(len(zahlen), sum(zahlen), max(zahlen))
```

## 2. Tupel

Wie Listen, aber unveränderbar (*immutable*) – gut für feste Wertepaare.

```python
punkt = (3, 7)
print(punkt[0], punkt[1])
# punkt[0] = 5   # würde einen Fehler geben
```

## 3. Dictionaries

```python
student = {"name": "Anna", "alter": 23, "note": 1.7}
print(student["name"])
student["note"] = 1.3           # Wert ändern
student["uni"] = "THI"          # neuen Schlüssel hinzufügen

for schluessel, wert in student.items():
    print(schluessel, "->", wert)
```

## 4. Sets

Ungeordnete Sammlung ohne doppelte Werte.

```python
fruechte = {"apfel", "banane", "apfel", "kirsche"}
print(fruechte)   # "apfel" erscheint nur einmal
print("banane" in fruechte)
```

## 5. List Comprehensions

Kurzschreibweise, um aus einer Liste eine neue zu erzeugen.

```python
zahlen = [1, 2, 3, 4, 5, 6]
quadrate = [z ** 2 for z in zahlen]
gerade = [z for z in zahlen if z % 2 == 0]
print(quadrate)
print(gerade)
```
