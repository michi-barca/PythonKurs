# Kapitel 1: Basics

Grundlagen: Variablen, Datentypen, Operatoren und Ein-/Ausgabe.

## 1. Variablen & Datentypen

```python
name = "Anna"        # str
alter = 23            # int
groesse = 1.72         # float
ist_student = True     # bool

print(type(name), type(alter), type(groesse), type(ist_student))
```

## 2. Rechenoperatoren

```python
a, b = 7, 2
print(a + b, a - b, a * b, a / b)   # / gibt immer float zurück
print(a // b)  # ganzzahlige Division
print(a % b)   # Rest (Modulo)
print(a ** b)  # Potenz
```

## 3. Strings

```python
vorname = "Anna"
nachname = "Meier"
voller_name = vorname + " " + nachname
print(voller_name)
print(f"{vorname} ist {alter} Jahre alt")   # f-String
print(voller_name.upper())
print(len(voller_name))
```

## 5. Typumwandlung

```python
zahl_text = "42"
zahl = int(zahl_text)
print(zahl + 8)

x = 3.9
print(int(x))   # schneidet Nachkommastellen ab, rundet nicht!
```
