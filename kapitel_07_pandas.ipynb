# Kapitel 10: Pandas Grundlagen

Pandas ist die Standard-Bibliothek für tabellarische Daten in Python – vergleichbar
mit einer Excel-Tabelle, aber programmierbar. Zentrales Objekt: der `DataFrame`.

Installation (falls nötig): `pip install pandas`

```python
import pandas as pd
```

## 1. DataFrame erstellen

Am einfachsten aus einem Dictionary: die Keys werden zu Spaltennamen.

```python
daten = {
    "name": ["Anna", "Ben", "Clara", "David"],
    "alter": [23, 25, 22, 28],
    "note": [1.7, 2.3, 1.3, 2.7]
}
df = pd.DataFrame(daten)
df
```

## 2. Erste Übersicht

```python
print(df.shape)     # (Zeilen, Spalten)
print(df.columns)   # Spaltennamen
df.head(2)           # erste 2 Zeilen
df.describe()        # Statistik über numerische Spalten
```

## 3. Spalten & Zeilen auswählen

```python
print(df["name"])          # eine Spalte -> Series
print(df[["name", "note"]]) # mehrere Spalten -> DataFrame

print(df.iloc[0])           # erste Zeile (nach Position)
print(df.loc[df["alter"] > 23])  # Zeilen mit Bedingung
```

## 4. Filtern

Genau wie bei List Comprehensions filtert man mit einer Bedingung –
nur eben spaltenweise statt elementweise.

```python
gute_noten = df[df["note"] < 2.0]
print(gute_noten)
```

## 5. Neue Spalte berechnen

```python
df["bestanden"] = df["note"] <= 4.0
df
```

## 6. Gruppieren mit `groupby`

```python
daten2 = {
    "fach": ["Mathe", "Mathe", "Info", "Info", "Info"],
    "note": [2.0, 1.3, 1.7, 2.3, 3.0]
}
df2 = pd.DataFrame(daten2)
print(df2.groupby("fach")["note"].mean())   # Durchschnittsnote je Fach
```

## 7. CSV lesen & schreiben

```python
df.to_csv("studenten.csv", index=False)
df_neu = pd.read_csv("studenten.csv")
df_neu
```
