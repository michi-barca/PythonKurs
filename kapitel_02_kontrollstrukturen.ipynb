# Kapitel 2: Kontrollstrukturen

Bedingungen und Schleifen – damit euer Programm Entscheidungen treffen und
Dinge wiederholen kann.

## 1. if / elif / else

```python
note = 2.3

if note <= 1.5:
    print("Sehr gut")
elif note <= 2.5:
    print("Gut")
elif note <= 4.0:
    print("Bestanden")
else:
    print("Nicht bestanden")
```

## 2. Vergleichs- & Logikoperatoren

```python
alter = 20
hat_ausweis = True

print(alter >= 18 and hat_ausweis)   # beide müssen wahr sein
print(alter < 18 or hat_ausweis)     # mindestens eine Bedingung
print(not hat_ausweis)
```

## 3. for-Schleife

```python
for i in range(5):
    print(i)   # 0,1,2,3,4

for zahl in [10, 20, 30]:
    print(zahl * 2)
```

## 4. while-Schleife

```python
zaehler = 0
while zaehler < 5:
    print(zaehler)
    zaehler += 1
```

## 5. break & continue

```python
for i in range(10):
    if i == 5:
        break        # Schleife komplett verlassen
    if i % 2 == 0:
        continue     # diesen Durchlauf überspringen
    print(i)
```
