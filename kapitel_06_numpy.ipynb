# Kapitel 9: NumPy Grundlagen

NumPy ist die Basis-Bibliothek für numerisches Rechnen in Python und die Grundlage
für pandas, scikit-learn & Co. Der große Vorteil: Rechenoperationen auf ganzen
Arrays sind viel schneller als Python-Schleifen über Listen.

Installation (falls nötig): `pip install numpy`

```python
import numpy as np
```

## 1. Arrays erstellen

```python
a = np.array([1, 2, 3, 4, 5])
print(a)
print(type(a))
print(a.shape)   # Form des Arrays
print(a.dtype)   # Datentyp der Elemente

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix.shape)  # (2, 3) -> 2 Zeilen, 3 Spalten
```

## 2. Nützliche Erzeugungsfunktionen

```python
print(np.zeros(5))          # 5 Nullen
print(np.ones((2, 3)))      # 2x3 Matrix mit Einsen
print(np.arange(0, 10, 2))  # 0, 2, 4, 6, 8
print(np.linspace(0, 1, 5)) # 5 gleich verteilte Werte zwischen 0 und 1
```

## 3. Indexing & Slicing – funktioniert wie bei Listen

```python
a = np.array([10, 20, 30, 40, 50])
print(a[0])      # 10
print(a[1:3])    # [20 30]
print(a[-1])     # 50

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix[0, 2])   # Zeile 0, Spalte 2 -> 3
print(matrix[:, 1])   # ganze Spalte 1 -> [2 5]
```

## 4. Vektorisierte Operationen

Das ist der eigentliche Grund, warum man NumPy statt Listen benutzt:
Operationen wirken automatisch auf **alle Elemente gleichzeitig** – ohne Schleife.

```python
a = np.array([1, 2, 3, 4])
print(a + 10)      # jedes Element +10
print(a * 2)       # jedes Element *2
print(a ** 2)      # jedes Element zum Quadrat

b = np.array([10, 20, 30, 40])
print(a + b)       # elementweise Addition
```

## 5. Broadcasting

NumPy kann auch mit Arrays unterschiedlicher Form rechnen, solange sie
kompatibel sind – der kleinere Wert wird automatisch "erweitert".

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix + 10)          # 10 wird auf jede Zelle addiert
print(matrix + np.array([1, 0, 1]))  # wird auf jede Zeile angewendet
```

## 6. Nützliche Statistik-Funktionen

```python
a = np.array([4, 8, 15, 16, 23, 42])
print(a.sum())
print(a.mean())
print(a.max(), a.min())
print(a.std())   # Standardabweichung
```
